### Importing packages

```python
import pandas as pd
import pickle
```

```python
# Load real FAQ dataset
df = pd.read_csv("codebasics_faqs.csv",encoding="cp1252")

# Inspect
print(df.head())
print(df.columns)
```

```python
# Convert rows to text documents
documents = (
    "Prompt: " + df["prompt"].astype(str) +
    ". Response: " + df["response"].astype(str)
).tolist()

print(f"Total documents: {len(documents)}")
print("\nSample document:\n", documents[0])
```

```python
# Save artifact
with open("documents.pkl", "wb") as f:
    pickle.dump(documents, f)

print("\nSaved documents.pkl")
```
