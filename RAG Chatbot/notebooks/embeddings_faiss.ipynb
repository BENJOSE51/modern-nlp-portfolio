```python
import os

BASE_DIR = os.getcwd()  # this notebook's folder

FAISS_PATH = os.path.join(BASE_DIR, "faiss.index")
CHUNKS_PATH = os.path.join(BASE_DIR, "chunks.txt")
```

```python
import pickle
import faiss
from sentence_transformers import SentenceTransformer
import numpy as np
```

```python
# Load chunks
with open("chunks.pkl", "rb") as f:
    chunks = pickle.load(f)

print(f"Loaded {len(chunks)} chunks")

# Embedding model
embed_model = SentenceTransformer("all-MiniLM-L6-v2")

# Encode chunks
embeddings = embed_model.encode(chunks).astype("float32")
print("Embedding shape:", embeddings.shape)
```

```python
# Build FAISS index
dimension = embeddings.shape[1]
index = faiss.IndexFlatL2(dimension)
index.add(embeddings)

print("Total vectors indexed:", index.ntotal)
```

```python
# Save FAISS index
faiss.write_index(index, "faiss.index")

# Save chunks for retrieval
with open("chunks.txt", "w", encoding="utf-8") as f:
    for c in chunks:
        f.write(c + "\n")

print("\nSaved faiss.index and chunks.txt")
```
